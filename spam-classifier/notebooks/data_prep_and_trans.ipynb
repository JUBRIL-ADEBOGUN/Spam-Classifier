```python
# Initialize a simple text vectorizer (e.g., TfidfVectorizer or CountVectorizer from scikit-learn).
# Create a function build_pipeline(X_train) that fits and returns the full transformation pipeline (Vectorizers, Scalers, etc.).)
def build_pipeline(X_train):
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.pipeline import Pipeline
    from sklearn.preprocessing import StandardScaler

    pipeline = Pipeline([
        ('tfidf', TfidfVectorizer()),
        ('scaler', StandardScaler(with_mean=False))
    ])
    
    pipeline.fit(X_train)
    return pipeline

```
